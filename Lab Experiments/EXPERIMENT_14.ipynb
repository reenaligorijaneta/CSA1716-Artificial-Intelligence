```python

def alphabeta(depth, node, maximizing, alpha, beta):
    if depth == 2:
        return values[node]

    if maximizing:
        best = -999
        for i in range(2):
            best = max(best, alphabeta(depth + 1, node * 2 + i,
                                       False, alpha, beta))
            alpha = max(alpha, best)
            if beta <= alpha:
                break
        return best

    else:
        best = 999
        for i in range(2):
            best = min(best, alphabeta(depth + 1, node * 2 + i,
                                       True, alpha, beta))
            beta = min(beta, best)
            if beta <= alpha:
                break
        return best

values = [3, 5, 2, 9]

print("Best value:", alphabeta(0, 0, True, -999, 999))
```
