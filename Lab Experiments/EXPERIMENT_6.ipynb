```python

room = ["Dirty", "Dirty"]
position = 0

while "Dirty" in room:
    if room[position] == "Dirty":
        print("Suck")
        room[position] = "Clean"

    if position == 0:
        print("Move Right")
        position = 1
    else:
        print("Move Left")
        position = 0

print("Both rooms are clean!")
```
