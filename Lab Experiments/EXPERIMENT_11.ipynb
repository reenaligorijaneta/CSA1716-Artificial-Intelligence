```python

colors = ['Red', 'Green', 'Blue']

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D'],
    'D': ['B', 'C']
}

color = {}

for region in graph:
    for c in colors:
        safe = True

        for neighbour in graph[region]:
            if color.get(neighbour) == c:
                safe = False

        if safe:
            color[region] = c
            break

print("Map Coloring:")
for region in color:
    print(region, ":", color[region])
```
