```python

from itertools import permutations

cost = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

cities = [1, 2, 3]
min_cost = 9999

for path in permutations(cities):
    total = cost[0][path[0]]

    for i in range(len(path) - 1):
        total += cost[path[i]][path[i + 1]]

    total += cost[path[-1]][0]

    if total < min_cost:
        min_cost = total
        best_path = path

print("Shortest path: A", end=" ")
for city in best_path:
    print(chr(city + 65), end=" ")
print("A")
print("Minimum cost:", min_cost)
```
