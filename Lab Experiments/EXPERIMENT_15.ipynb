```python

from sklearn.tree import DecisionTreeClassifier

# Training data
X = [[1], [2], [3], [4], [5], [6]]
y = [0, 0, 0, 1, 1, 1]

# Create and train the decision tree
model = DecisionTreeClassifier()
model.fit(X, y)

# Predict the result
result = model.predict([[4]])

print("Prediction:", result[0])
```
