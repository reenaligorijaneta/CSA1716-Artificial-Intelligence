```python

graph = {
    'A': [('B', 1), ('C', 3)],
    'B': [('D', 1), ('E', 5)],
    'C': [('F', 2)],
    'D': [],
    'E': [('G', 2)],
    'F': [('G', 1)],
    'G': []
}

h = {'A': 4, 'B': 3, 'C': 2, 'D': 2, 'E': 1, 'F': 1, 'G': 0}

open_list = [(h['A'], 0, 'A')]
visited = []

while open_list:
    open_list.sort()
    f, cost, node = open_list.pop(0)

    if node in visited:
        continue

    visited.append(node)
    print(node, end=" ")

    if node == 'G':
        print("\nGoal reached!")
        break

    for neighbour, distance in graph[node]:
        if neighbour not in visited:
            new_cost = cost + distance
            open_list.append((new_cost + h[neighbour], new_cost, neighbour))
```
