```python

from collections import deque

queue = deque([(3, 3, 0)])
visited = set()

while queue:
    m, c, b = queue.popleft()

    if (m, c, b) in visited:
        continue
    visited.add((m, c, b))

    print("Left:", m, c, "Right:", 3-m, 3-c)

    if m == 0 and c == 0:
        print("All crossed safely!")
        break

    for x, y in [(1,0), (2,0), (0,1), (0,2), (1,1)]:
        if b == 0:
            nm, nc, nb = m-x, c-y, 1
        else:
            nm, nc, nb = m+x, c+y, 0

        if 0 <= nm <= 3 and 0 <= nc <= 3:
            if (nm == 0 or nm >= nc) and (3-nm == 0 or 3-nm >= 3-nc):
                if (nm, nc, nb) not in visited:
                    queue.append((nm, nc, nb))
```
